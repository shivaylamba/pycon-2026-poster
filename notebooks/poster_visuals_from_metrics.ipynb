## Poster Heuristic Table

This table gives one compact row per workload and hardware profile for use in the poster narrative.

```python
heuristics = df.groupby(['workload', 'hardware_label']).agg(
    mean_latency_s=('total_latency_s', 'mean'),
    mean_tokens=('total_tokens', 'mean'),
    mean_cost_usd=('total_cost_usd', 'mean'),
    mean_energy_j=('total_energy_j', 'mean'),
    mean_energy_per_1k_tokens=('energy_j_per_1k_tokens', 'mean'),
).reset_index()
heuristics
```
